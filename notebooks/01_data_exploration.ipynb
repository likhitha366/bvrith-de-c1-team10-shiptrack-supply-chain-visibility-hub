## Mission
Explore source datasets, inspect schemas, validate grains and prepare a Week-3 Bronze demonstration table.

```python
from pyspark.sql import SparkSession
spark=SparkSession.builder.getOrCreate()
volume_path="/Volumes/ship_track/default/ship_track_volume"
```

## Explore `shipments.parquet`

```python
import pyarrow.parquet as pq
import pandas as pd

# Read with PyArrow to handle nanosecond timestamps
parquet_table = pq.read_table(f"{volume_path}/shipments.parquet")
pandas_df = parquet_table.to_pandas()

# Convert to Spark DataFrame
df_shipments = spark.createDataFrame(pandas_df)
display(df_shipments)
df_shipments.printSchema()
df_shipments.createOrReplaceTempView("shipments")
```

```python
spark.sql('SELECT * FROM shipments LIMIT 10').show()
```

## Explore `hubs.json`

```python
df_hubs=spark.read.format("json")\
    .option("multiLine", "true")\
    .load(f"{volume_path}/hubs.json")
display(df_hubs)
df_hubs.printSchema()
df_hubs.createOrReplaceTempView("hubs")
```

```python
spark.sql('SELECT * FROM hubs LIMIT 10').show()
```

## Bronze Demonstration Table

```python
%sql
CREATE OR REPLACE TABLE workspace.default.shiptrack_week03_bronze_demo_shipments AS
SELECT * FROM shipments;
```

```python
%sql
DESCRIBE DETAIL workspace.default.shiptrack_week03_bronze_demo_shipments;
DESCRIBE HISTORY workspace.default.shiptrack_week03_bronze_demo_shipments;
```

## Lineage Demonstration View

```python
%sql
CREATE OR REPLACE VIEW workspace.default.shiptrack_week03_lineage_demo_view AS
SELECT shipment_id, latest_status FROM shiptrack_week03_bronze_demo_shipments;
```
