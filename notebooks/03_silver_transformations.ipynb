# 03 — Silver Transformations

**Week:** 5

Goal: Standardize raw Bronze data into clean Silver table(s).


### Table 1 - Scan Events

```python
bronze_df = spark.table("bronze_scan_events")

silver_df = bronze_df.selectExpr(
    "source_record_id AS record_id",
    "shipment_id AS reference_id",
    "CAST(event_time AS timestamp) AS event_timestamp",
    "CAST(to_date(event_time) AS date) AS event_date",
    "event_type",
    "LOWER(shipment_status) AS status_standard",
    "source_system",
    "ingestion_timestamp",
    "scan_id",
    "event_sequence_no",
    "hub_id",
    "carrier_id",
    "route_id",
    "exception_type",
    "package_condition",
    "status_reason",
    "run_id",
    "source_file_name",
    "batch_id"
)

display(silver_df.limit(10))

silver_df.write.mode("overwrite").saveAsTable("silver_scan_events")

print("Rows:", spark.table("silver_scan_events").count())
```

### Table 2 - Shipments

```python
bronze_df = spark.table("bronze_shipments")

silver_df = bronze_df.dropDuplicates()

display(silver_df.limit(10))

silver_df.write.mode("overwrite").saveAsTable("silver_shipments")

print("Rows:", spark.table("silver_shipments").count())
```
